```python
import numpy as np
import talib as ta
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
#imports aapl ticker as "aapl"
aapl= yf.Ticker("AAPL")
# downloads  historical data as from five years as it provides a good balance, capturing a few different economic cycles
aapl_data= yf.download("AAPL", start="2020-01-01", end=None)

#feature engineering/ foundations of modal
#EMA to mesure trend over a span of 50
aapl_data['EMA_50']= aapl_data[['Close']].ewm(span=50).mean()

#RSI to mesure the the momentum/reversal
aapl_data['rsi_50'] = ta.RSI(aapl_data[('Close', 'AAPL')],timeperiod=50)

#bollinger bands to mesure the volatility
upperband, middleband, lowerband= ta.BBANDS(aapl_data[('Close','AAPL')], timeperiod=50, nbdevup=2, nbdevdn=2, matype=0)
aapl_data['bb_upper'] = upperband
aapl_data['bb_middle'] = middleband
aapl_data['bb_lower'] = lowerband


#plots historical data and moving average on a graph
aapl_data[('Close')].plot(label='Close')
aapl_data['EMA_50'].plot(label='EMA 50')
aapl_data['rsi_50'].plot(label='RSI')
aapl_data['bb_upper'].plot(label='upper band')
aapl_data['bb_middle'].plot(label='middle band')
aapl_data['bb_lower'].plot(label='lower band')
plt.title("Apple Stock Prices")
plt.legend()
plt.show()
aapl_data
```
